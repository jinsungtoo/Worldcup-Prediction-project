### 팀별 선수 시장 가치 비교

```python
# 팀별 스쿼드&선수 이름 파일 가져오기
file_path1 =  '/content/drive/MyDrive/World cup Project/archive2/squads_and_players.csv'
squads_and_players = pd.read_csv(file_path1)
display(squads_and_players)

# 팀별 파일 가져오기
file_path2 = '/content/drive/MyDrive/World cup Project/archive2/teams.csv'
teams = pd.read_csv(file_path2)

## 팀별 선수 시장 가치 비교
market_value = (squads_and_players
                .groupby("team_id",as_index=False)["market_value_eur"]
                .sum())

## 내림차순으로 결과 출력
result = (
    pd.merge(market_value,
                  teams[["team_id", "team_name"]],
                  on="team_id",
                  how="left"
                  )
    [["team_id", "team_name", "market_value_eur"]]
        .sort_values(by="market_value_eur", ascending=False)
        .reset_index(drop=True)
)


## 천 단위 "," 표시와 "유로" 단위 표시
display(result.style.format({"market_value_eur": "{:,.0f} 유로"
    })
)
```

#### 선수가치가 높은 팀일수록 월드컵에서 좋은 성적을 거둘까?

```python
#승점을 기준으로 각 팀의 경기 기록 만들기

file_path3 = '/content/drive/MyDrive/World cup Project/archive2/matches.csv'
matches = pd.read_csv(file_path3)

## 홈팀 데이터
home = matches[
    ["home_team_id", "home_score", "away_score"]
].copy()

home = home.rename(columns={
    "home_team_id": "team_id",
    "home_score": "goals_for",
    "away_score": "goals_against"
})

## 원정팀 데이터
away = matches[
    ["away_team_id", "away_score", "home_score"]
].copy()

away = away.rename(columns={
    "away_team_id": "team_id",
    "away_score": "goals_for",
    "home_score": "goals_against"
})

## 하나로 합치기
team_matches = pd.concat([home, away], ignore_index=True)

display(team_matches)

print('-'*80)

# 승 무 패 계산
team_matches["result"] = "D"

team_matches.loc[
    team_matches["goals_for"] > team_matches["goals_against"],
    "result"
] = "W"

team_matches.loc[
    team_matches["goals_for"] < team_matches["goals_against"],
    "result"
] = "L"

## 팀별 집계
team_performance = (
    team_matches
    .groupby("team_id")
    .agg(
        경기수=("result", "count"),
        승리=("result", lambda x: (x == "W").sum()),
        무승부=("result", lambda x: (x == "D").sum()),
        패배=("result", lambda x: (x == "L").sum()),
        득점=("goals_for", "sum"),
        실점=("goals_against", "sum")
    )
    .reset_index()
)

## 승률 승점 추가
team_performance["승률"] = (
    team_performance["승리"] /
    team_performance["경기수"]
)

team_performance["승점"] = (
    team_performance["승리"] * 3 +
    team_performance["무승부"]
)
# plt.scatter(
#     result["market_value_eur"],
#     result["win_rate"]
# )

# plt.xlabel("Team Market Value")
# plt.ylabel("Win Rate")
# plt.title("Market Value vs Win Rate")

# plt.show()
```

##### 시장가치와 합치기

```python
analysis = pd.merge(
    result,
    team_performance,
    on="team_id",
    how="inner"
)

# 팀 이름 정의
analysis = analysis[
    [
        "team_id",
        "team_name",
        "market_value_eur",
        "경기수",
        "승리",
        "무승부",
        "패배",
        "득점",
        "실점",
        "승률",
        "승점"
    ]
]
```

##### 시장가치와 승률의 상관관계

```python
analysis[
    ["market_value_eur", "승률"]
].corr()

# 즉, 팀의 총 선수 시장가치가 높을수록 75%라는 높은 양의 상관관계를 가지고 있다고 볼 수 있다.
```

## 2026 우승팀 예측

### 머신러닝 예측 모델 학습

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier

# 1. 팀별 피처(Feature) 생성 (시장가치 + 최근 승률/득실점 지표)
# 1-1. analysis 데이터프레임 기반으로 기초 데이터 준비
team_stats = analysis.copy()

# 경기당 평균 득점 및 실점 추가
team_stats['avg_goals_for'] = team_stats['득점'] / team_stats['경기수']
team_stats['avg_goals_against'] = team_stats['실점'] / team_stats['경기수']

# 예측에 사용할 핵심 Feature 수집
features_df = team_stats[['team_id', 'team_name', 'market_value_eur', '승률', 'avg_goals_for', 'avg_goals_against']].set_index('team_id')


# 2. 머신러닝 학습 데이터셋 구축 (matches 기반)
train_data = []

for idx, row in matches.iterrows():
    h_id = row['home_team_id']
    a_id = row['away_team_id']

    # 두 팀이 모두 features_df에 존재하는 경우만 학습
    if h_id in features_df.index and a_id in features_df.index:
        home_feat = features_df.loc[h_id]
        away_feat = features_df.loc[a_id]

        # 두 팀의 전력 차이(Diff)를 Feature로 구성
        val_diff = home_feat['market_value_eur'] - away_feat['market_value_eur']
        win_rate_diff = home_feat['승률'] - away_feat['승률']
        gf_diff = home_feat['avg_goals_for'] - away_feat['avg_goals_for']
        ga_diff = home_feat['avg_goals_against'] - away_feat['avg_goals_against']

        # 타겟 변수 (1: 홈승, 0: 무승부, -1: 원정승)
        if row['home_score'] > row['away_score']:
            target = 1
        elif row['home_score'] == row['away_score']:
            target = 0
        else:
            target = -1

        train_data.append([val_diff, win_rate_diff, gf_diff, ga_diff, target])

# 학습용 데이터프레임 생성
train_df = pd.DataFrame(train_data, columns=['val_diff', 'win_rate_diff', 'gf_diff', 'ga_diff', 'target'])

X_train = train_df[['val_diff', 'win_rate_diff', 'gf_diff', 'ga_diff']]
y_train = train_df['target']


# 3. 모델 학습 (RandomForest)
model = RandomForestClassifier(n_estimators=100, random_state=0)
model.fit(X_train, y_train)

print("✅ 승/무/패 예측 모델 학습 완료!")
```

### 경기 결과 예측 함수 & 시뮬레이션

```python
# 4. 단판 경기 승자 예측 함수 (토너먼트용: 무승부 시 승부차기 확률 처리)
def predict_match_winner(team_a_id, team_b_id):
    # 두 팀의 데이터 조회
    feat_a = features_df.loc[team_a_id]
    feat_b = features_df.loc[team_b_id]

    # 두 팀 간 전력 차이 계산
    diff_vector = np.array([[
        feat_a['market_value_eur'] - feat_b['market_value_eur'],
        feat_a['승률'] - feat_b['승률'],
        feat_a['avg_goals_for'] - feat_b['avg_goals_for'],
        feat_a['avg_goals_against'] - feat_b['avg_goals_against']
    ]])

    # [원정승(-1) 확률, 무승부(0) 확률, 홈승(1) 확률]
    probs = model.predict_proba(diff_vector)[0]
    classes = model.classes_

    prob_b = probs[np.where(classes == -1)[0][0]] if -1 in classes else 0
    prob_draw = probs[np.where(classes == 0)[0][0]] if 0 in classes else 0
    prob_a = probs[np.where(classes == 1)[0][0]] if 1 in classes else 0

    # 토너먼트 단판 승부: 무승부 발생 시 승률 비율로 승부차기 승자 결정
    sim_val = np.random.rand()
    if sim_val < prob_a:
        return team_a_id
    elif sim_val < prob_a + prob_b:
        return team_b_id
    else:
        # 무승부일 경우 전력이 약간 높은 쪽에 가중치를 두어 승부차기 승자 결정
        p_a_penalty = prob_a / (prob_a + prob_b + 1e-5)
        return team_a_id if np.random.rand() < p_a_penalty else team_b_id


# 5. 몬테카를로 시뮬레이션 (1000회 실행)
def run_simulation(num_simulations=1000):
    all_teams = list(features_df.index)
    win_counts = {t_id: 0 for t_id in all_teams}

    # 시뮬레이션 용이성을 위해 상위 32개 팀 선택 (시장가치 순)
    top_32_teams = features_df.sort_values(by='market_value_eur', ascending=False).head(32).index.tolist()

    for _ in range(num_simulations):
        current_round = top_32_teams.copy()
        np.random.shuffle(current_round) # 토너먼트 대진 랜덤 셔플

        # 32강 -> 16강 -> 8강 -> 4강 -> 결승
        while len(current_round) > 1:
            next_round = []
            for i in range(0, len(current_round), 2):
                team_1 = current_round[i]
                team_2 = current_round[i+1]
                winner = predict_match_winner(team_1, team_2)
                next_round.append(winner)
            current_round = next_round

        final_champion = current_round[0]
        win_counts[final_champion] += 1

    # 결과 정렬 및 데이터프레임 생성
    results = []
    for t_id, count in win_counts.items():
        if t_id in top_32_teams:
            t_name = features_df.loc[t_id, 'team_name']
            prob = (count / num_simulations) * 100
            results.append({'Team': t_name, 'Win_Probability (%)': round(prob, 2)})

    res_df = pd.DataFrame(results).sort_values(by='Win_Probability (%)', ascending=False).reset_index(drop=True)
    return res_df

# 시뮬레이션 실행 및 출력
df_worldcup_prediction = run_simulation(num_simulations=1000)
display(df_worldcup_prediction.head(10))
```
